```python
import os
import pandas as pd
emails_path = "C:/Users/Owner/OneDrive/Desktop/Phising/data/raw/emails/"  # Adjust if needed

# List all CSV files in the emails directory
email_files = [f for f in os.listdir(emails_path) if f.endswith('.csv')]

# Load and combine all email datasets
all_emails_df = pd.concat([pd.read_csv(os.path.join(emails_path, file)) for file in email_files], ignore_index=True)

# Display dataset information
print("Combined Emails Dataset Info:")
print(all_emails_df.info())
```

```python
print("Email Data Preview:")
print(all_emails_df.head())

```

```python
all_emails_df['subject'].fillna('No Subject', inplace=True)
all_emails_df['body'].fillna('No Content', inplace=True)
all_emails_df['urls'].fillna('', inplace=True)
all_emails_df.drop(columns=['sender', 'receiver', 'date'], inplace=True)

print(all_emails_df.isnull().sum())
```
